```python
import pandas as pd

stats = pd.read_csv("../data/epl_player_stats_24_25.csv")
print(stats.shape)
print(stats.columns.tolist())
stats.head()
```

```python
stats.info()
stats.isna().sum().sort_values(ascending=False).head(10)
```

```python
pct_cols = [c for c in stats.columns if "%" in c]
print(pct_cols)

for c in pct_cols:
    stats[c] = stats[c].astype(str).str.replace("%", "", regex=False).astype(float)

print("Duplicate rows:", stats.duplicated().sum())
print("Clubs:", stats["Club"].nunique())
print(stats["Position"].value_counts())
```

```python
bad = stats[stats["Shots On Target"] > stats["Shots"]]
print(len(bad), "rows where shots on target > shots")
print(bad[["Player Name", "Club", "Shots", "Shots On Target"]].head(10))

print(stats["Minutes"].describe())
print("Players under 450 minutes:", (stats["Minutes"] < 450).sum())
```

```python
print(stats["Club"].value_counts())
```

```python
stats["Club"] = stats["Club"].replace({"Brighton": "Brighton & Hove Albion"})
print(stats["Club"].nunique())   # should print 20
```

```python
print(stats[["Shots", "Shots On Target"]].corr())

print(stats.nlargest(10, "Shots On Target")[["Player Name", "Position", "Shots", "Shots On Target"]])

check = stats[stats["Shots"] > 0].copy()
check["calc_conv"] = (check["Goals"] / check["Shots"] * 100).round(1)
print(check[["Player Name", "Goals", "Shots", "Conversion %", "calc_conv"]].head(10))
```

```python
stats = stats.drop(columns=["Shots On Target", "Conversion %"])

# (part, total): part should never exceed total
pairs = [
    ("Successful Passes", "Passes"),
    ("Successful Crosses", "Crosses"),
    ("Successful fThird Passes", "fThird Passes"),
    ("gDuels Won", "Ground Duels"),
    ("aDuels Won", "Aerial Duels"),
    ("Progressive Carries", "Carries"),
    ("Carries Ended with Goal", "Carries"),
    ("Carries Ended with Shot", "Carries"),
    ("Goals", "Shots"),
    ("Appearances", "Appearances"),
]

for part, total in pairs:
    n_bad = (stats[part] > stats[total]).sum()
    print(f"{part} > {total}: {n_bad} rows")
```

```python
calc = stats["Successful Passes"] / stats["Passes"] * 100
print((calc - stats["Passes%"]).abs().describe())

calc = stats["gDuels Won"] / stats["Ground Duels"] * 100
print((calc - stats["gDuels %"]).abs().describe())
```

```python
print("rows in stats:", len(stats))
print("zero passes:", (stats["Passes"] == 0).sum())
print("zero ground duels:", (stats["Ground Duels"] == 0).sum())
print("percentages that are NaN:", stats["Passes%"].isna().sum())
```

```python
zero = stats["Passes"] == 0

# 1. Are these just low-minute players?
print("Zero-pass players by minutes:")
print(stats[zero]["Minutes"].describe())
print("Zero-pass with 900+ minutes:", (zero & (stats["Minutes"] >= 900)).sum())

# 2. Is it concentrated in certain clubs?
print(stats.groupby("Club")["Passes"].apply(lambda s: (s == 0).mean()).sort_values(ascending=False).round(2))

# 3. Are other columns empty too, or just passes/duels?
print(stats[zero][["Touches", "Tackles", "Carries", "Shots", "Goals", "Assists"]].describe().loc[["mean", "max"]])

# 4. Look at a few examples
print(stats[zero][["Player Name", "Club", "Position", "Minutes", "Goals", "Touches", "Passes"]].head(10))
```

```python
zero = stats["Passes"] == 0
num_cols = stats.select_dtypes("number").columns.drop(["Appearances", "Minutes"])

# 1. Which columns form the missing block?
rep = pd.DataFrame({
    "zero_in_missing_group": (stats.loc[zero, num_cols] == 0).mean(),
    "zero_in_other_group": (stats.loc[~zero, num_cols] == 0).mean(),
}).round(2)
print(rep.sort_values("zero_in_missing_group", ascending=False).head(25))
```

```python
# 2. How many usable players would we have?
print(pd.crosstab(stats["Minutes"] >= 450, zero, rownames=["450+ min"], colnames=["missing block"]))
print(pd.crosstab(stats["Position"], zero))
```

```python
# 3. Does missingness depend on the name (the join hypothesis)?
nonascii = stats["Player Name"].apply(lambda s: not s.isascii())
print(pd.crosstab(nonascii, zero, normalize="index").round(2))
```

```python
zero = stats["Passes"] == 0

print("Complete players, minutes:")
print(stats.loc[~zero, "Minutes"].describe())

miss450 = stats[zero & (stats["Minutes"] >= 450)]
print("\nMissing but 450+ minutes, by position:")
print(miss450["Position"].value_counts())

print(miss450.sort_values("Minutes", ascending=False)[
    ["Player Name", "Club", "Position", "Minutes"]].head(15))
```

```python
outfield = stats[stats["Position"] != "GKP"]

print("Outfield, 900+ min:", (outfield["Minutes"] >= 900).sum())
print("...of which complete:", ((outfield["Minutes"] >= 900) & (outfield["Passes"] > 0)).sum())
print("Min minutes among complete outfield:", outfield.loc[outfield["Passes"] > 0, "Minutes"].min())
print("Max minutes among missing outfield:", outfield.loc[outfield["Passes"] == 0, "Minutes"].max())

print(stats[(stats["Position"] == "GKP") & (stats["Passes"] > 0)][["Player Name", "Club", "Minutes"]])
```

```python
out = stats[stats["Position"] != "GKP"]
print(out[(out["Minutes"] >= 900) & (out["Passes"] == 0)][["Player Name", "Club", "Position", "Minutes"]])
```
